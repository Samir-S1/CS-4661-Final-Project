Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
```

Load the data

```python
print("Loading CSV...")
df = pd.read_csv("../data/CompleteData/AirNow_AQS_Meteo_Joined_Jan2020-Oct2025.csv", dtype=str)
print(f"Loading Complete")
```

Find all unique sensors with their particle.

```python
cols = df.columns.tolist()
sensors = df[[cols[0], cols[1], cols[3], cols[6]]].drop_duplicates()
sensors.columns = ['lat', 'lon', 'particle', 'site']
sensors['lat'] = pd.to_numeric(sensors['lat'])
sensors['lon'] = pd.to_numeric(sensors['lon'])
print(f"Number of sensors: {len(sensors)}")
```

Plot sensors and make the map

```python
particles = sensors['particle'].unique().tolist()

# Render PM2.5 on top
particles.remove('PM2.5')
particles.append('PM2.5')

fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(1, 1, 1, projection=ccrs.PlateCarree())

ax.add_feature(cfeature.LAND, facecolor='lightgray')
ax.add_feature(cfeature.OCEAN, facecolor='lightblue')
ax.add_feature(cfeature.COASTLINE, linewidths=0.5)
ax.add_feature(cfeature.BORDERS, linewidths=0.5)
ax.add_feature(cfeature.STATES, linewidths=0.5)

for particle in particles:
    data = sensors[sensors['particle'] == particle]
    ax.scatter(data['lon'], data['lat'],
               label=particle, alpha=1.0, edgecolors='k', linewidths=0.3,
               transform=ccrs.PlateCarree(), zorder=5)

ax.set_title('Sensor Locations in California')
ax.legend(loc='lower left')
ax.gridlines(draw_labels=True, alpha=0.4)
plt.show()
```
